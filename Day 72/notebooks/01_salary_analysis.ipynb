## 2. Useful Pandas Methods

| Method or Attribute | Purpose |
|:---|:---|
| `df.shape` | Returns the number of rows and columns |
| `df.shape[0]` | Returns the number of rows |
| `df.shape[1]` | Returns the number of columns |
| `df.columns` | Displays the column labels |
| `df.info()` | Shows column names, data types, and non-null counts |
| `df.isna().sum()` | Counts missing values in each column |
| `df.describe()` | Summarizes numerical columns |
| `df.duplicated().sum()` | Counts duplicate rows |
