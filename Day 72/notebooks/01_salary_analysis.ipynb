# College Major Salary Analysis

Explore the salary dataset with Pandas. This notebook inspects the raw data, checks its quality, cleans the incomplete row, and answers basic questions about starting and mid-career pay. Run the notebook from this folder (Day 72/notebooks) so the relative data paths resolve.


```python
import pandas as pd
from IPython.display import display

raw_path = "../data/raw/salaries_by_college_major.csv"
df = pd.read_csv(raw_path)
display(df.head())
```

## 1. Inspect the dataset

Check its dimensions, column names, data types, missing values, and duplicate rows before analyzing it. The raw file includes an incomplete final row, which should be removed before calculations.


```python
print(f"Rows: {df.shape[0]}")
print(f"Columns: {df.shape[1]}")
print("Column names:", df.columns.to_list())
df.info()
```

```python
display(df.isna().sum().rename("Missing values"))
print(f"Duplicate rows: {df.duplicated().sum()}")
display(df[df.isna().any(axis=1)])
```

## 2. Clean and summarize

Drop rows with missing values, then review numeric summaries and the number of majors in each category. Keep the raw DataFrame unchanged so the cleaning step is easy to compare.


```python
clean_df = df.dropna().copy()
print(f"Rows after removing incomplete data: {len(clean_df)}")
display(clean_df.describe().round(2))
display(clean_df["Group"].value_counts().rename("Number of majors"))
```

## 3. Compare majors by salary

Find the majors with the highest and lowest starting median salaries and mid-career median salaries. Sorting makes the comparisons explicit and easy to adjust.


```python
salary_columns = ["Starting Median Salary", "Mid-Career Median Salary"]
for column in salary_columns:
    print(f"Highest {column.lower()}")
    display(clean_df.nlargest(5, column)[["Undergraduate Major", "Group", column]])
    print(f"Lowest {column.lower()}")
    display(clean_df.nsmallest(5, column)[["Undergraduate Major", "Group", column]])
```

## 4. Explore a category

Filter the data to one group and compare its starting and mid-career salary medians. Change selected_group to STEM, Business, or HASS to explore another group.


```python
selected_group = "STEM"
group_df = clean_df.loc[clean_df["Group"] == selected_group]
display(group_df[["Undergraduate Major", *salary_columns]].sort_values("Mid-Career Median Salary", ascending=False))
```
