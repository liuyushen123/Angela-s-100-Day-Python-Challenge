# Salary Growth and Career Risk

Compare salary growth from the start of a career to mid-career, and use the gap between the 90th and 10th mid-career salary percentiles as a simple measure of salary spread. A wider spread means outcomes vary more; it is not a guarantee of personal career risk.


```python
import pandas as pd

raw_path = "../data/raw/salaries_by_college_major.csv"
df = pd.read_csv(raw_path).dropna().copy()
```

## 1. Calculate growth and spread

Growth is expressed as a percentage of the starting median salary. Spread is the 90th percentile minus the 10th percentile at mid-career.


```python
df["Growth %"] = (
    (df["Mid-Career Median Salary"] - df["Starting Median Salary"])
    / df["Starting Median Salary"] * 100
)
df["Spread"] = (
    df["Mid-Career 90th Percentile Salary"]
    - df["Mid-Career 10th Percentile Salary"]
)
display(df[["Undergraduate Major", "Starting Median Salary", "Mid-Career Median Salary", "Growth %", "Spread"]].head())
```

## 2. Find the largest and smallest spreads

The majors at the top of this ranking have the widest distance between the reported 90th and 10th percentile salaries. The smallest spreads indicate a narrower range in this dataset.


```python
risk_columns = [
    "Undergraduate Major", "Group",
    "Mid-Career 10th Percentile Salary",
    "Mid-Career 90th Percentile Salary", "Spread",
]
print("Widest mid-career salary spreads")
display(df.nlargest(10, "Spread")[risk_columns])
print("Narrowest mid-career salary spreads")
display(df.nsmallest(10, "Spread")[risk_columns])
```

## 3. Compare growth

Sort by growth percentage to see which majors have the largest relative increase from starting to mid-career pay. Also check whether any majors have a negative change.


```python
growth_columns = [
    "Undergraduate Major", "Group", "Starting Median Salary",
    "Mid-Career Median Salary", "Growth %",
]
print("Largest percentage growth")
display(df.nlargest(10, "Growth %")[growth_columns])
print("Negative mid-career change")
display(df.loc[df["Growth %"] < 0, growth_columns])
```

## 4. Inspect a specific major

Use a case-insensitive search to compare the starting, mid-career, and percentile salaries for majors whose names contain a term of interest.


```python
search_term = "engineering"
major_matches = df[df["Undergraduate Major"].str.contains(search_term, case=False, na=False)]
display(major_matches[[
    "Undergraduate Major", "Starting Median Salary",
    "Mid-Career Median Salary", "Mid-Career 10th Percentile Salary",
    "Mid-Career 90th Percentile Salary", "Growth %", "Spread",
]].sort_values("Mid-Career Median Salary", ascending=False))
```
