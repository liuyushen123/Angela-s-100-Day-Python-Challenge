### `ndim` Attribute

The `ndim` attribute in NumPy returns the **number of dimensions (axes)** of an ndarray.

In simple terms, it tells us how many dimensions an array has.

**Examples:**

\`\`\`python
import numpy as np

# 1-dimensional array

arr_1d = np.array([1, 2, 3, 4])
print(arr_1d.ndim) # Output: 1

# 2-dimensional array

arr_2d = np.array([[1, 2, 3], [4, 5, 6]])
print(arr_2d.ndim) # Output: 2

# 3-dimensional array

arr_3d = np.array([
[[1, 2], [3, 4]],
[[5, 6], [7, 8]]
])
print(arr_3d.ndim) # Output: 3
\`\`\`

**Key Takeaways:**

- `ndim` tells us the number of dimensions in an array.
- 1D = A single row or sequence of values.
- 2D = A matrix with rows and columns.
- 3D = A collection of matrices.
- `ndim` is an attribute, not a method, so we use `arr.ndim`, not `arr.ndim()`.

**Remember:** `ndim` tells us _how many dimensions_ an array has, not how many elements it contains.

#### 2-Dimensional Arrays (Matrices)


### NumPy Array Shape Requirements

Unlike regular Python Lists, NumPy's multidimensional arrays generally require a **regular (homogeneous) shape**.

This means that each nested list must contain the same number of elements when creating a standard multidimensional `ndarray`.

**Example – Invalid Array:**

```python
import numpy as np

array_2d = np.array([
    [1, 2, 3, 9],     # 4 elements
    [5, 6, 7, 8, 2]   # 5 elements
])
```

This raises a `ValueError` because the nested lists have different lengths.

**Example – Valid Array:**

```python
array_2d = np.array([
    [1, 2, 3, 9],
    [5, 6, 7, 8]
])

print(array_2d.shape)  # (2, 4)
```

**Key Takeaways:**

- Each row must have the same number of elements.
- A regular 2D array has a consistent number of rows and columns.
- `shape` tells us the size of each dimension.
- `ndim` tells us how many dimensions the array has.

**Remember:** NumPy arrays are designed for efficient numerical computation, so maintaining a consistent structure is important.


````python
array_2d = np.array([[1, 2, 3, 9], [5, 6, 7, 8]])
````

````python
print(f"array_2d has {array_2d.ndim} dimensions")
````

````python
print(f"it's shape is {array_2d.shape}")
````

````python
print(f"It has {array_2d.shape[0]} rows and {array_2d.shape[1]} columns")
````

### Finding Elements and Their Coordinates in NumPy

NumPy provides `np.where()` and `np.argwhere()` to locate elements that satisfy a given condition.

**1. Using `np.where()`**

```python
coordinates = np.where(array_3d == 10)
```

Returns a tuple containing the indices along each axis.

**2. Using `np.argwhere()`**

```python
coordinates = np.argwhere(array_3d == 10)
```

Returns an array where each row represents the coordinates of a matching element.

**Key Takeaways:**

- `np.where()` returns separate index arrays for each dimension.
- `np.argwhere()` returns complete coordinate tuples as rows.
- For a 3D array, each coordinate contains three indices: `(axis_0, axis_1, axis_2)`.
- You can use these coordinates to access the original element.


````python
np.argwhere(mystery_array == 18)
````

````python
target_array = [97, 0, 27, 18]

coordinates = mystery_array == target_array
print(coordinates)
````

````python
result = np.any(coordinates, axis=2)
print(result)
````

````python
coordinate = np.argwhere(result)
print(f"Target Array Location: {coordinate[0][0]}, {coordinate[0][1]}")
````
